````python
import pandas as pd
import requests
````

````python
def get_teams_for_season_id(season_id):
  teams_response = requests.get("https://lscluster.hockeytech.com/feed/index.php?feed=modulekit&view=teamsbyseason&season_id="+ str(season_id)+"&key=446521baf8c38984&client_code=pwhl")
  if teams_response.status_code != 200:
    return None

  teams_json = teams_response.json()
  teams_list = teams_json["SiteKit"]["Teamsbyseason"]

  # Load the JSON file into a DataFrame
  teams_df = pd.json_normalize(teams_list)

  return teams_df
````

````python
def get_roster_for_season_id_and_team_id(season_id, team_id):
  roster_response = requests.get("https://lscluster.hockeytech.com/feed/index.php?feed=modulekit&view=roster&team_id="+str(team_id)+"&season_id="+str(season_id)+"&key=446521baf8c38984&client_code=pwhl")
  if roster_response.status_code != 200:
    return None

  roster_json = roster_response.json()
  roster_list = roster_json["SiteKit"]["Roster"]
  roster_without_coaches = [item for item in roster_list if not isinstance(item, list)]

  # Load the JSON file into a DataFrame
  roster_df = pd.json_normalize(roster_without_coaches)

  return roster_df
````

````python
def get_team_rosters_for_season_id(season_id):
  teams_df = get_teams_for_season_id(season_id)
  team_dict = {}
  player_id_map_for_season = {} # New: to store player name to ID mapping for this season

  for _, team in teams_df.iterrows():
    roster_df = get_roster_for_season_id_and_team_id(season_id, team["id"])
    if roster_df is not None and not roster_df.empty: # Check if roster_df is not None and not empty
        roster_list = set(roster_df["name"].tolist())
        # Populate player_id_map_for_season
        # Iterate over the roster_df to get both 'name' and 'person_id'
        for _, player_row in roster_df.iterrows():
            player_name = player_row["name"]
            player_id = player_row["id"]
            player_id_map_for_season[player_name] = player_id
    else:
        roster_list = set() # If no roster or empty, set an empty set

    city_name = team["city"]
    if city_name == "Montreal":
      city_name = "Montréal"
    team_dict[city_name] = roster_list
  return team_dict, player_id_map_for_season

def get_all_team_rosters():
  season_team_roster_dict = {}
  all_players_id_map = {} # New: to store player name to ID mapping across all seasons

  for season_name in season_to_id_map.keys():
    # Call the modified get_team_rosters_for_season_id
    team_rosters, player_id_map_for_season = get_team_rosters_for_season_id(season_to_id_map[season_name])
    season_team_roster_dict[season_name] = team_rosters
    all_players_id_map.update(player_id_map_for_season) # Update the master map

  return season_team_roster_dict, all_players_id_map
````

# Player source target dicts (without stats)
Given two seasons, create a player_source_target_dict which maps a player to their source-target transition in between seasons
```
{'Amanda Pelkey': ('Boston 2024', 'Boston 2024-2025'),
 'Jessica DiGirolamo': ('Boston 2024', 'Boston 2024-2025'),
 'Susanna Tapani': ('Minnesota 2024', 'Boston 2024-2025'),
 'Emily Brown': ('Boston 2024', 'Boston 2024-2025'),
 ```

player_source_target_dicts_list is a list of the player_source_target_dicts with sources 2024, 2024-2025, and 2025-2026.
```
 player_source_target_dicts_list = [player_source_target_2024, player_source_target_2025, player_source_target_2026]
 ```

````python
def generate_source_target_player_dict(begin_season, end_season):
  begin_rosters = all_team_rosters[begin_season]
  end_rosters = all_team_rosters[end_season]

  player_source_target_dict = {}

  for begin_team in begin_rosters.keys():
    begin_team_roster = begin_rosters[begin_team]
    for player in begin_team_roster:
      for end_team in end_rosters.keys():
        end_team_roster = end_rosters[end_team]
        if player in end_team_roster:
          source_target = (begin_team +" "+ begin_season, end_team +" "+ end_season)
          player_source_target_dict[player]=source_target

  return player_source_target_dict
````

````python
def get_goals_for_player_by_season(player_name):
  player_id = player_id_map[player_name]

  player_stats_response = requests.get("https://lscluster.hockeytech.com/feed/index.php?feed=modulekit&view=player&category=seasonstats&player_id="+player_id+"&key=446521baf8c38984&client_code=pwhl")
  if player_stats_response.status_code != 200:
    return None

  season_stats_dict = {}

  player_stats_json = player_stats_response.json()
  all_season_stats = player_stats_json["SiteKit"]["Player"]
  if isinstance(all_season_stats, bool):
    # print(player_name)
    return None

  regular_season_stats = all_season_stats.get("regular")
  if regular_season_stats is None:
    print(player_name)
    return None

  for season_stats in regular_season_stats:
    if season_stats["season_name"] != "Total":
      season_stats_dict[season_stats["season_id"]] = season_stats["goals"]

  return season_stats_dict
````

````python
from collections import defaultdict
import itertools
````

````python
# Generate player count, player lists, goals for every source, target tuple
source_target_count = defaultdict(int)
source_target_player_lists = defaultdict(list)
source_target_goals = defaultdict(int)

source_target_info = defaultdict(dict)


for player_source_target_dict in player_source_target_dicts_list_goals:
    for player_name, source_target_goals_tuple in player_source_target_dict.items():
        source_target_tuple, goals = source_target_goals_tuple

        if source_target_info.get(source_target_tuple) is None:
            source_target_info[source_target_tuple] = {"player_count": 0, "player_list": [], "goals":0}


        source_target_info[source_target_tuple]["player_count"] +=1
        source_target_info[source_target_tuple]["player_list"].append(player_name)
        source_target_info[source_target_tuple]["goals"] += int(goals)

        source_target_count[source_target_tuple] += 1
        source_target_player_lists[source_target_tuple].append(player_name)
        source_target_goals[source_target_tuple] += int(goals)
````

````python
import altair as alt
````

````python
# Define custom colors based on the provided JavaScript with full city names
custom_colors = {
    "Seattle": "teal",
    "Toronto": "RoyalBlue",
    "Boston": "green",
    "Montréal": "maroon",
    "Hamilton": "gold",
    "Detroit": "black",
    "Las Vegas": "Olive",
    "San Jose": "DeepSkyBlue",
    "Minnesota": "MediumSlateBlue",
    "Ottawa": "Crimson",
    "Vancouver": "MediumBlue",
    "New York": "LightSeaGreen",
}

# Create lists for domain (team names) and range (colors)
color_domain = list(custom_colors.keys())
color_range = list(custom_colors.values())
````

````python
def create_bar_chart(transition_df, vertical_label):

  # Filter for transitions originating from the 2025-2026 season
  df_2025_2026_transitions = transition_df[transition_df['source'].str.contains('2025-2026')].copy()

  # Extract source and target team names (full city names)
  df_2025_2026_transitions['source_team'] = df_2025_2026_transitions['source'].apply(lambda x: ' '.join(x.split(' ')[:-1]))
  df_2025_2026_transitions['target_team'] = df_2025_2026_transitions['target'].apply(lambda x: ' '.join(x.split(' ')[:-1]))

  # Calculate players lost by each 2025-2026 team
  lost_players_by_team = {}

  # Get all unique source teams for 2025-2026 season
  source_teams_2025_2026 = df_2025_2026_transitions['source'].unique()

  for team_full_name in source_teams_2025_2026:
      source_team_name = ' '.join(team_full_name.split(' ')[:-1]) # e.g., 'Boston' or 'New York'

      # Filter transitions for the current source team
      team_transitions = df_2025_2026_transitions[df_2025_2026_transitions['source'] == team_full_name]

      # Sum players where the target team is different from the source team
      total_lost = team_transitions[team_transitions['target_team'] != source_team_name]['value'].sum()
      lost_players_by_team[source_team_name] = total_lost

  # print("Number of players lost by each 2025-2026 team:")
  # for team, count in lost_players_by_team.items():
  #     print(f"{team}: {count} players lost")


  # Define a fixed sort order for target teams for consistent visualization
  target_team_sort_order = {
      'Detroit': 1,
      'Hamilton': 2,
      'Las Vegas': 3,
      'San Jose': 4,
      'Boston': 5,
      'Minnesota': 6,
      'Montréal': 7,
      'New York': 8,
      'Ottawa': 9,
      'Seattle': 10,
      'Toronto': 11,
      'Vancouver': 12,
  }

  # Create a sorting key to put the original team at the bottom, then expansion teams, then others
  df_2025_2026_transitions['sort_order'] = df_2025_2026_transitions.apply(
      lambda row: 0 if row['source_team'] == row['target_team'] else target_team_sort_order.get(row['target_team'], len(target_team_sort_order) + 1),
      axis=1
  )

  # Create the stacked bar graph with custom colors and sorting
  chart = alt.Chart(df_2025_2026_transitions).mark_bar().encode(
      x=alt.X('source_team:N', axis=alt.Axis(title='Source Team')),
      y=alt.Y('value:Q', axis=alt.Axis(title=vertical_label)),
      color=alt.Color('target_team:N', title='Target Team',
                      scale=alt.Scale(domain=color_domain, range=color_range)),
      order=alt.Order('sort_order:Q', sort='ascending'), # Sort by the new column
      tooltip=['source_team', 'target_team', 'value', alt.Tooltip('player_list', title='Players')]
  ).properties(
      title='Player Transitions from Source Teams (by Target Team)'
  )

  # for source_team in lost_players_by_team.keys():
  #   print(f"\nPlayers lost by {source_team}:")

  #   # Filter for transitions from this source team to a different target team
  #   lost_transitions = df_2025_2026_transitions[
  #       (df_2025_2026_transitions['source_team'] == source_team) &
  #       (df_2025_2026_transitions['target_team'] != source_team)
  #   ]

  #   if lost_transitions.empty:
  #       print("  No players lost to other teams.")
  #   else:
  #       for index, row in lost_transitions.iterrows():
  #           target_team = row['target_team']
  #           players = row['player_list']
  #           if players:
  #               print(f"  To {target_team}: {', '.join(players)}")

  return chart
````

````python
def create_expansion_target_bar_chart(transition_df, vertical_label):
  # Define the target expansion teams
  expansion_target_teams = ['Hamilton', 'Detroit', 'Las Vegas', 'San Jose']

  # Filter for transitions *to* the 2026-2027 season
  df_2026_2027_targets = transition_df[transition_df['target'].str.contains('2026-2027')].copy()

  # Extract source and target team names (full city names)
  df_2026_2027_targets['source_team'] = df_2026_2027_targets['source'].apply(lambda x: ' '.join(x.split(' ')[:-1]))
  df_2026_2027_targets['target_team'] = df_2026_2027_targets['target'].apply(lambda x: ' '.join(x.split(' ')[:-1]))

  # Further filter to only include transitions to the specified expansion teams
  df_expansion_targets = df_2026_2027_targets[
      df_2026_2027_targets['target_team'].isin(expansion_target_teams)
  ].copy()

  # Define a consistent order for the x-axis (target expansion teams)
  target_team_x_order = [team for team in expansion_target_teams if team in df_expansion_targets['target_team'].unique()]

  # Create the stacked bar chart with custom colors
  chart = alt.Chart(df_expansion_targets).mark_bar().encode(
      # X-axis: Target expansion teams
      x=alt.X('target_team:N', axis=alt.Axis(title='Target Expansion Team'), sort=target_team_x_order),
      # Y-axis: Value (player count or goals)
      y=alt.Y('value:Q', axis=alt.Axis(title=vertical_label)),
      # Color: Source teams
      color=alt.Color('source_team:N', title='Source Team',
                      scale=alt.Scale(domain=color_domain, range=color_range)),
      tooltip=[
          alt.Tooltip('source_team', title='Source Team'),
          alt.Tooltip('target_team', title='Target Team'),
          alt.Tooltip('value:Q', title=vertical_label),
          alt.Tooltip('player_list', title='Players')
      ]
  ).properties(
      title=f'Expansion Team Makeup - {vertical_label}'
  )

  return chart
````

# Vertical Diverging Stacked Bar Chart: Value Kept vs. Lost (2025-2026 Season)

````python
def diverging_stacked_bars(transition_df, title, ascending=False):
  # Filter for transitions originating from the 2025-2026 season
  df_2025_2026_transitions = transition_df[transition_df['source'].str.contains('2025-2026')].copy()

  # Extract source and target team names (full city names)
  df_2025_2026_transitions['source_team'] = df_2025_2026_transitions['source'].apply(lambda x: ' '.join(x.split(' ')[:-1]))
  df_2025_2026_transitions['target_team'] = df_2025_2026_transitions['target'].apply(lambda x: ' '.join(x.split(' ')[:-1]))

  # Create a copy to avoid modifying the original DataFrame used by the first chart
  df_diverging = df_2025_2026_transitions.copy()

  # Create a 'display_value' for the diverging chart:
  # Positive for players kept, negative for players lost
  df_diverging['display_value'] = df_diverging.apply(
      lambda row: row['value'] if row['source_team'] == row['target_team'] else -row['value'],
      axis=1
  )

  # Calculate players lost by each 2025-2026 team (for sorting)
  value_lost = df_diverging[df_diverging['source_team'] != df_diverging['target_team']].groupby('source_team')['value'].sum().reset_index()
  value_lost.rename(columns={'value': 'players_lost'}, inplace=True)

  # Get the order of source teams based on total players lost
  sorted_source_teams_by_lost = value_lost.sort_values(by='players_lost', ascending=ascending)['source_team'].tolist()

  # Define a fixed sort order for target teams for consistent visualization (from create_bar_chart)
  target_team_sort_order = {
      'Detroit': 1,
      'Hamilton': 2,
      'Las Vegas': 3,
      'San Jose': 4,
      'Boston': 5,
      'Minnesota': 6,
      'Montréal': 7,
      'New York': 8,
      'Ottawa': 9,
      'Seattle': 10,
      'Toronto': 11,
      'Vancouver': 12,
  }

  # Create a sorting key to put the original team at the bottom, then expansion teams, then others
  df_diverging['sort_order'] = df_diverging.apply(
      lambda row: 0 if row['source_team'] == row['target_team'] else target_team_sort_order.get(row['target_team'], len(target_team_sort_order) + 1),
      axis=1
  )

  # Define the base chart with a vertical orientation
  base_vertical = alt.Chart(df_diverging).encode(
      x=alt.X('source_team:N',
              axis=alt.Axis(title='Source Team', labelAngle=-45, titleFontSize=20, labelFontSize=18),
              sort=sorted_source_teams_by_lost), # Sort x-axis by players lost
      tooltip=['source_team', 'target_team', 'value', alt.Tooltip('player_list', title='Players')] # Keep original 'value' for tooltip
  )

  # Create the stacked bars
  vertical_bar_chart = base_vertical.mark_bar().encode(
      y=alt.Y('display_value:Q',
              axis=alt.Axis(title=title + ' (Kept: Positive, Lost: Negative)', format='d', titleFontSize=20, labelFontSize=18)),
      color=alt.Color('target_team:N', title='Target Team',
                      scale=alt.Scale(domain=color_domain, range=color_range),
                      legend=alt.Legend(titleFontSize=20, labelFontSize=18)), # Increase legend title and label font sizes
      # Use the same sort_order to maintain consistent stacking within each bar
      # For the diverging chart, this sorts from the central axis outwards.
      order=alt.Order('sort_order:Q', sort='ascending')
  ).properties(
      title=alt.Title(text=[title +' Kept vs Lost For Existing', 'PWHL Teams in the 2026 Expansion'], anchor='middle', fontSize=24),
      width=400, # Restored width to 400
      height=500  # Restored height to 500
  )

  return vertical_bar_chart
````
